## What was the most frequent movie duration in the 1990s?

```python
# Printing first five rows of films that was released at 1990
movies_1990s = netflix_df[
    (netflix_df['type'] == 'Movie') & 
    (netflix_df['release_year'] >= 1990) & 
    (netflix_df['release_year'] < 2000)
]

# Saving films all duration as an integer all_duration. Also making it numpy array
display(movies_1990s.head())

# Finding and printing moda through all durations
duration = movies_1990s['duration'].mode()[0]

print(f"The most frequnet movie duration is - {duration}")
```

## Count the number of short action movies released in the 1990s.

```python
short_movie_count = len(movies_1990s[
    (movies_1990s['genre'] == 'Action') &
    (movies_1990s['duration'] < 90)
])
print(short_movie_count)
```
